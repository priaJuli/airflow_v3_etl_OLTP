# Hands-On: Data Exploration

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, upper, when, count_distinct

# 1. Initialize the connection to the Master
spark = SparkSession.builder \
    .appName("Jupyter_to_Spark_Master") \
    .master("spark://spark-master:7077") \
    .config("spark.executor.memory", "4g") \
    .config("spark.executor.cores", "2") \
    .config("spark.cores.max", "6") \
    .config("spark.sql.shuffle.partitions", "12") \
    .getOrCreate()
```

```python
import time

start = time.perf_counter()
# ... your code here ...

path_to_file = '/home/jovyan/datasets/synthetic_healthcare_semantics.csv'
df = spark.read.csv(path_to_file, header=True)

end = time.perf_counter()

print(f"Execution time: {end - start:.6f} seconds")
```

```python
df.show(5)
```

```python
print(len(df.columns))
end = time.perf_counter()

print(f"Execution time: {end - start:.6f} seconds")
```

```python
df.printSchema()
```

```python

print(df.describe().toPandas())
end = time.perf_counter()

print(f"Execution time: {end - start:.6f} seconds")
```
